# Markov Chain Simulation

This notebook demonstrates a discrete-time Markov chain simulation for modeling state transitions.

## Overview

A Markov chain is a stochastic model describing a sequence of possible events where the probability of each event depends only on the state in the previous event. This implementation simulates a traveler moving between cities based on transition probabilities.

## Example Scenario

We model a traveler visiting three cities: **Berlin**, **Munich**, and **Hamburg**. The transition probabilities are defined in a transition matrix, where each row represents the current city and each column represents the probability of moving to each city (including staying in the same city).

### Transition Matrix

```
         Berlin  Munich  Hamburg
Berlin    0.5     0.3     0.2
Munich    0.4     0.4     0.2
Hamburg   0.1     0.4     0.5
```

The simulation runs for 100 steps, starting from Berlin, and tracks how many times the traveler visits each city.
